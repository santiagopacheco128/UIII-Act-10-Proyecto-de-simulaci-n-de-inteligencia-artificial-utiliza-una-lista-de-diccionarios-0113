```python
print("🎭 SISTEMA INTELIGENTE DE ACTORES 🎭")

# Lista vacía
actor = []

# Diccionario vacío
losactores = {}

while True:

    print("\n========= MENÚ PRINCIPAL =========")
    print("1. Registrar actor")
    print("2. Mostrar actores")
    print("3. Buscar actor")
    print("4. Modificar actor")
    print("5. Eliminar actor")
    print("6. Salir")

    opcion = input("Selecciona una opción: ")

    # OPCIÓN 1
    if opcion == "1":

        nombre = input("Nombre del actor: ")
        edad = input("Edad: ")
        genero = input("Género favorito (Acción, Drama, Comedia): ")
        experiencia = input("Años de experiencia: ")

        losactores = {
            "nombre": nombre,
            "edad": edad,
            "genero": genero,
            "experiencia": experiencia
        }

        actor.append(losactores)

        print("\n✅ Actor registrado correctamente")

    # OPCIÓN 2
    elif opcion == "2":

        if len(actor) == 0:
            print("\n⚠️ No hay actores registrados")

        else:
            print("\n🎬 LISTA DE ACTORES")

            contador = 1

            for datos in actor:

                print("\nActor", contador)
                print("Nombre:", datos["nombre"])
                print("Edad:", datos["edad"])
                print("Género:", datos["genero"])
                print("Experiencia:", datos["experiencia"])

                contador += 1

    # OPCIÓN 3
    elif opcion == "3":

        buscar = input("Ingresa el nombre del actor a buscar: ")

        encontrado = False

        for datos in actor:

            if datos["nombre"] == buscar:

                print("\n🎭 ACTOR ENCONTRADO")
                print("Nombre:", datos["nombre"])
                print("Edad:", datos["edad"])
                print("Género:", datos["genero"])
                print("Experiencia:", datos["experiencia"])

                encontrado = True

        if encontrado == False:
            print("\n❌ Actor no encontrado")

    # OPCIÓN 4
    elif opcion == "4":

        modificar = input("Nombre del actor a modificar: ")

        encontrado = False

        for datos in actor:

            if datos["nombre"] == modificar:

                print("\n✏️ Ingresa los nuevos datos")

                datos["edad"] = input("Nueva edad: ")
                datos["genero"] = input("Nuevo género: ")
                datos["experiencia"] = input("Nueva experiencia: ")

                print("\n✅ Actor modificado correctamente")

                encontrado = True

        if encontrado == False:
            print("\n❌ Actor no encontrado")

    # OPCIÓN 5
    elif opcion == "5":

        eliminar = input("Nombre del actor a eliminar: ")

        encontrado = False

        for datos in actor:

            if datos["nombre"] == eliminar:

                actor.remove(datos)

                print("\n🗑️ Actor eliminado correctamente")

                encontrado = True

        if encontrado == False:
            print("\n❌ Actor no encontrado")

    # OPCIÓN 6
    elif opcion == "6":

        print("\n👋 Gracias por usar el sistema")
        print("Programa realizado por Santiago Ernesto Pacheco Holguin")
        break

    # OPCIÓN INVÁLIDA
    else:
        print("\n⚠️ Opción inválida")
```
